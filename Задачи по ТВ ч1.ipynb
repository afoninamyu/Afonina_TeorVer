### Пример 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

а). Количество вариантов распределения **различных** призов:

${\widetilde{A}_{10}^5} = 10^5 = 100000$

```python
from itertools import product

films = '0123456789'

a = set(product(films, repeat=5))
len(a)
```

б). Количество вариантов распределения **одинаковых** призов:

${\widetilde{С}_{10}^5} = {С_{10+5-1}^5} = {С_{14}^5} = \frac{14 \cdot 13 \cdot 12 \cdot 11 \cdot 10}{1 \cdot 2 \cdot 3 \cdot 4 \cdot 5} = 2002$

```python
sc.comb(10, 5, exact=True, repetition=True)
```

### Пример 1.10

$P(A) = \frac{m}{n} = \frac{1}{A_{6}^5} = \frac{1}{6 \cdot 5 \cdot 4} = \frac{1}{120}$

```python
from itertools import permutations

num = 'ТЕОРИЯ'

a = set(permutations(num, 3))
round(1 / len(a), 4)
```

```python
round(1 / sc.perm(6, 3), 4)
```

б). Рассмотрим событие $B$ – получение слова «ТЕОРИЯ».

$P(B) = \frac{m}{n} = \frac{1}{P_6} = \frac{1}{6!} = \frac{1}{720}$

```python
from itertools import permutations

num = 'ТЕОРИЯ'

a = list(permutations(num))
round(1 / len(a), 4)
```

### Пример 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

a). Рассмотрим событие А - никакие два лица одного пола не сядут рядом.

$P(A) = \frac{m}{n} = \frac{2(5!)^2}{10!} = \frac{2 \cdot 120^2}{3628800} = 0,00794$

### Пример 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

a). Рассмотрим событие А - пассажиры попали в два купе.

$P(A) = \frac{m}{n} = \frac{C_9^2 \cdot C_8^7}{C_{36}^7} = \frac{36 \cdot 8}{8347680} = 0,0000345$

б). Рассмотрим событие В - пассажиры попали в семь купе.

$P(B) = \frac{m}{n} = \frac{C_9^7 \cdot 4^7}{C_{36}^7} = \frac{36 \cdot 16384}{8347680} = 0,07066$

```python
round(sc.comb(9, 7) * 4**7 / sc.comb(36, 7), 5)
```

$P(D) = \frac{m}{n} = \frac{C_9^3(6 C_4^4 C_4^2 C_4^1 + 3 C_4^3 C_4^3 C_4^1 + 3 C_4^3 C_4^2 C_4^2)}{C_{36}^7} = \frac{64512}{8347680} = 0,00773$

```python
round(sc.comb(9, 3) * (6*sc.comb(4, 4)*sc.comb(4, 2)*sc.comb(4, 1) + 3*sc.comb(4, 3)*sc.comb(4, 3)*sc.comb(4, 1) + 3*sc.comb(4, 3)*sc.comb(4, 2)*sc.comb(4, 2)) / sc.comb(36, 7), 5)
```

### Задача 1.48

> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

а). Рассмотрим событие $A$ – при бросании трех игральных костей в сумме выпало 11 очков

```python
from itertools import product

ls = list(product(range(1,7), repeat=3))

cnt = 0
for el in ls:
    if sum(el) == 11:
        cnt += 1
cnt / len(ls)
```

б). Рассмотрим событие $B$ – при бросании трех игральных костей в сумме выпало более 10 очков

```python
cnt = 0
for el in ls:
    if sum(el) > 10:
        cnt += 1
cnt / len(ls)
```

# 4. Условная вероятность

> Решить классически и смоделировать решение задачи. В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

Рассмотрим события: $H_i$ – среди вынутых из 1-ой урны 3-х шаров i белых; A - шар, вынутый из 2-ой урны, белый.

$P(A) = P(H_0)\cdot P_{H_0}(A) + P(H_1)\cdot P_{H_1}(A) + P(H_2)\cdot P_{H_2}(A) + P(H_3)\cdot P_{H_3}(A) = \frac{C_5^3\cdot C_{15}^0}{C_{20}^3}\cdot \frac{6+0}{10} + \frac{C_5^2\cdot C_{15}^1}{C_{20}^3}\cdot \frac{6+1}{10} + \frac{C_5^1\cdot C_{15}^2}{C_{20}^3}\cdot \frac{6+2}{10} + \frac{C_5^0\cdot C_{15}^3}{C_{20}^3}\cdot \frac{6+3}{10} = 0,825$

```python
round(sc.comb(5, 3) * sc.comb(15, 0) / sc.comb(20, 3) * 6/10 + sc.comb(5, 2) * sc.comb(15, 1) / sc.comb(20, 3) * 7/10 + sc.comb(5, 1) * sc.comb(15, 2) / sc.comb(20, 3) * 8/10 + sc.comb(5, 0) * sc.comb(15, 3) / sc.comb(20, 3) * 9/10, 4)
```

```python
from random import sample, choice

def give_and_take():
    urn1 = ["W"] * 15 + ["B"] * 5
    urn2 = ["W"] * 6 + ["B"]

    give = sample(urn1, 3)
    urn2.extend(give)
    take = choice(urn2)

    return take == "W"

cnt_white = sum(1 for i in range(1_000_000) if give_and_take())
round(cnt_white / 1_000_000, 4)
```

### Пример 2.9
> В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций в результате торгов по первоначально заявленной цене: 1) не будут проданы 5 пакетов; 2) будет продано: а) менее 2 пакетов; б) не более 2 пакетов; в) хотя бы 2 пакета; г) наивероятнейшее число пакетов.

```python
from scipy.stats import binom, poisson, norm
from math import sqrt
```

```python
n = 9
q = 0.2
p = 1 - q
m = 5
prob = binom.pmf(m, n, p)
print(f"1). Вероятность, что из {n} пакетов акций по первоначально заявленной цене не будут проданы {m} пакетов: {prob:.4f}.")
```

```python
p = 0.2
m = 1
prob = binom.cdf(m, n, p)
print(f"2, а). Вероятность, что из {n} пакетов акций по первоначально заявленной цене будет продано менее {m+1} пакетов: {prob:.4f}.")
```

```python
m = 2
prob = binom.cdf(m, n, p)
print(f"2, б). Вероятность, что из {n} пакетов акций по первоначально заявленной цене будет продано не более {m} пакетов: {prob:.4f}.")
```

```python
m = 1
prob = 1 - binom.cdf(m, n, p)
print(f"2, в). Вероятность, что из {n} пакетов акций по первоначально заявленной цене будет продано хотя бы {m+1} пакета: {prob:.4f}.")
```

```python
q = 1 - p
m0 = [mo for mo in range(n+1) if (n*p-q) <= mo <= (n*p+p)]
prob = binom.pmf(m0[0], n, p) + binom.pmf(m0[1], n, p)
print(f"2, г). Наивероятнейших чисел два: {m0[0]} и {m0[1]}. Вероятность, что из {n} пакетов акций по первоначально заявленной цене будет продано наивероятнейшее число пакетов.: {prob:.4f}.")
```

### Пример 2.10
> Завод отправил на базу 10 000 стандартных изделий. Среднее число изделий, повреждаемых при транспортировке, составляет 0,02%. Найти вероятность того, что из 10 000 изделий: 1) будет повреждено: а) 3; б) по крайней мере 3; 2) не будет повреждено: а) 9997; б) хотя бы 9997.

```python
n = 10_000
p = 0.0002
lmb = n * p
m = 3
prob = poisson.pmf(k=m, mu=lmb)
print(f"1, а). Вероятность, что из {n} изделий будет повреждено {m}: {prob:.4f}.")
```

```python
m = 2
prob = 1 - poisson.cdf(k=m, mu=lmb)
print(f"1, б). Вероятность, что из {n} изделий будет повреждено по крайней мере {m+1}: {prob:.4f}.")
```

```python
m = 3
prob = poisson.cdf(k=m, mu=lmb)
print(f"2, б). Вероятность, что из {n} изделий не будет повреждено хотя бы {n-m}: {prob:.4f}.")
```

### Пример 2.11
> По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины: а) 480 предприятий; б) наивероятнейшее число предприятий; в) не менее 480; г) от 480 до 520.

```python
n = 1000
p = 0.5
q = 1 - p
m = 480
x = (m - n * p) / sqrt(n * p * q)
prob = norm.pdf(x)/sqrt(n*p*q)
print(f"а). Вероятность, что из {n} предприятий {m} имеют нарушения финансовой дисциплины: {prob:.4f}.")
```

```python
m0 = [mo for mo in range(n+1) if (n*p-q) <= mo <= (n*p+p)]
x = (m0[0] - n * p) / sqrt(n * p * q)
prob = norm.pdf(x)/sqrt(n*p*q)
print(f"б). Вероятность, что из {n} предприятий имеет нарушения финансовой дисциплины наивероятнейшее число предприятий {m0[0]}: {prob:.4f}.")
```

```python
a = 480
x = (a - n * p) / sqrt(n * p * q)
prob = 1 - norm.cdf(x)
print(f"в). Вероятность, что из {n} предприятий не менее {a} имеют нарушения финансовой дисциплины: {prob:.4f}.")
```

```python
a = 480
b = 520
x1 = (a - n * p) / sqrt(n * p * q)
x2 = (b - n * p) / sqrt(n * p * q)
prob = norm.cdf(x2) - norm.cdf(x1)
print(f"г). Вероятность, что из {n} предприятий от {a} до {b} имеют нарушения финансовой дисциплины: {prob:.4f}.")
```
