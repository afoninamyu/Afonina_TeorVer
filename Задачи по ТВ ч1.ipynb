### Пример 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

а). Количество вариантов распределения **различных** призов:

${\widetilde{A}_{10}^5} = 10^5 = 100000$

```python
from itertools import product

films = '0123456789'

a = set(product(films, repeat=5))
len(a)
```

б). Количество вариантов распределения **одинаковых** призов:

${\widetilde{С}_{10}^5} = {С_{10+5-1}^5} = {С_{14}^5} = \frac{14 \cdot 13 \cdot 12 \cdot 11 \cdot 10}{1 \cdot 2 \cdot 3 \cdot 4 \cdot 5} = 2002$

```python
sc.comb(10, 5, exact=True, repetition=True)
```

### Пример 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

Количество искомых семизначных чисел:

$P_7(3; 2; 2) = \frac{17!}{3!2!2!} = 210$

```python
from itertools import permutations

num = '4445566'

a = set(permutations(num))
len(a)
```

```python
sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2))
```

### Пример 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

a). Рассмотрим событие $A$ – получение слова «ТОР»

$P(A) = \frac{m}{n} = \frac{1}{A_{6}^5} = \frac{1}{6 \cdot 5 \cdot 4} = \frac{1}{120}$

```python
from itertools import permutations

num = 'ТЕОРИЯ'

a = set(permutations(num, 3))
round(1 / len(a), 4)
```

```python
round(1 / sc.perm(6, 3), 4)
```

б). Рассмотрим событие $B$ – получение слова «ТЕОРИЯ».

$P(B) = \frac{m}{n} = \frac{1}{P_6} = \frac{1}{6!} = \frac{1}{720}$

```python
from itertools import permutations

num = 'ТЕОРИЯ'

a = list(permutations(num))
round(1 / len(a), 4)
```

```python
round(1 / sc.factorial(6), 4)
```

> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

Рассмотрим событие $B$ – получение слова «АНАНАС».

$P(B) = \frac{m}{n} = \frac{P_3 \cdot P_2}{P_6} = \frac{3!2!}{6!} = \frac{1}{60}$

```python
round(sc.factorial(3) * sc.factorial(2) / sc.factorial(6), 4)
```

### Пример 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

a). Рассмотрим событие А - никакие два лица одного пола не сядут рядом.

$P(A) = \frac{m}{n} = \frac{2(5!)^2}{10!} = \frac{2 \cdot 120^2}{3628800} = 0,00794$

```python
round(2 * (sc.factorial(5))**2 / sc.factorial(10), 5)
```

б). Рассмотрим событие В - супруги, образующие пять супружеских пар, сядут рядом.

$P(B) = \frac{m}{n} = \frac{5!2^5}{10!} = \frac{120 \cdot 32}{3628800} = 0,00106$

```python
round(sc.factorial(5) * 2**5 / sc.factorial(10), 5)
```

### Пример 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

a). Рассмотрим событие А - пассажиры попали в два купе.

$P(A) = \frac{m}{n} = \frac{C_9^2 \cdot C_8^7}{C_{36}^7} = \frac{36 \cdot 8}{8347680} = 0,0000345$

```python
round(sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7), 7)
```

б). Рассмотрим событие В - пассажиры попали в семь купе.

$P(B) = \frac{m}{n} = \frac{C_9^7 \cdot 4^7}{C_{36}^7} = \frac{36 \cdot 16384}{8347680} = 0,07066$

```python
round(sc.comb(9, 7) * 4**7 / sc.comb(36, 7), 5)
```

в). Рассмотрим событие D - пассажиры попали в три купе.

$P(D) = \frac{m}{n} = \frac{C_9^3(6 C_4^4 C_4^2 C_4^1 + 3 C_4^3 C_4^3 C_4^1 + 3 C_4^3 C_4^2 C_4^2)}{C_{36}^7} = \frac{64512}{8347680} = 0,00773$

```python
round(sc.comb(9, 3) * (6*sc.comb(4, 4)*sc.comb(4, 2)*sc.comb(4, 1) + 3*sc.comb(4, 3)*sc.comb(4, 3)*sc.comb(4, 1) + 3*sc.comb(4, 3)*sc.comb(4, 2)*sc.comb(4, 2)) / sc.comb(36, 7), 5)
```

### Задача 1.38
> Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

Рассмотрим событие $A$ – книги стоят слева направо в порядке нумерации томов

### Задача 1.43

а). Рассмотрим событие $A$ – все цифры в номере различные

$P(A) = \frac{m}{n} = \frac{0,9 \cdot A_{10}^5}{90000}$

```python
round(0.9 * sc.perm(10, 5) / 90000, 4)
```

б). Рассмотрим событие B – все цифры в номере одинаковые

$P(B) = \frac{m}{n} = \frac{9}{90000} = \frac{1}{10000} = 0,0001$

```python
9 / 90000
```

в). Рассмотрим событие C – все цифры в номере нечетные

$P(C) = \frac{m}{n} = \frac{5^5}{90000}$

```python
round(5**5 / 90000, 4)
```

### Задача 1.48

> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

а). Рассмотрим событие $A$ – при бросании трех игральных костей в сумме выпало 11 очков

```python
from itertools import product

ls = list(product(range(1,7), repeat=3))

cnt = 0
for el in ls:
    if sum(el) == 11:
        cnt += 1
cnt / len(ls)
```

б). Рассмотрим событие $B$ – при бросании трех игральных костей в сумме выпало более 10 очков

```python
cnt = 0
for el in ls:
    if sum(el) > 10:
        cnt += 1
cnt / len(ls)
```

# 4. Условная вероятность

> Решить классически и смоделировать решение задачи. В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

Рассмотрим события: $H_i$ – среди вынутых из 1-ой урны 3-х шаров i белых; A - шар, вынутый из 2-ой урны, белый.

$P(A) = P(H_0)\cdot P_{H_0}(A) + P(H_1)\cdot P_{H_1}(A) + P(H_2)\cdot P_{H_2}(A) + P(H_3)\cdot P_{H_3}(A) = \frac{C_5^3\cdot C_{15}^0}{C_{20}^3}\cdot \frac{6+0}{10} + \frac{C_5^2\cdot C_{15}^1}{C_{20}^3}\cdot \frac{6+1}{10} + \frac{C_5^1\cdot C_{15}^2}{C_{20}^3}\cdot \frac{6+2}{10} + \frac{C_5^0\cdot C_{15}^3}{C_{20}^3}\cdot \frac{6+3}{10} = 0,825$

```python
round(sc.comb(5, 3) * sc.comb(15, 0) / sc.comb(20, 3) * 6/10 + sc.comb(5, 2) * sc.comb(15, 1) / sc.comb(20, 3) * 7/10 + sc.comb(5, 1) * sc.comb(15, 2) / sc.comb(20, 3) * 8/10 + sc.comb(5, 0) * sc.comb(15, 3) / sc.comb(20, 3) * 9/10, 4)
```

```python
from random import sample, choice

def give_and_take():
    urn1 = ["W"] * 15 + ["B"] * 5
    urn2 = ["W"] * 6 + ["B"]

    give = sample(urn1, 3)
    urn2.extend(give)
    take = choice(urn2)

    return take == "W"

cnt_white = sum(1 for i in range(1_000_000) if give_and_take())
round(cnt_white / 1_000_000, 4)
```
