### Пример 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

а). Количество вариантов распределения **различных** призов:

${\widetilde{A}_{10}^5} = 10^5 = 100000$

```python
from itertools import product

films = '0123456789'

a = set(product(films, repeat=5))
len(a)
```

б). Количество вариантов распределения **одинаковых** призов:

${\widetilde{С}_{10}^5} = {С_{10+5-1}^5} = {С_{14}^5} = \frac{14 \cdot 13 \cdot 12 \cdot 11 \cdot 10}{1 \cdot 2 \cdot 3 \cdot 4 \cdot 5} = 2002$

```python
sc.comb(10, 5, exact=True, repetition=True)
```

### Пример 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

Количество искомых семизначных чисел:

$P_7(3; 2; 2) = \frac{17!}{3!2!2!} = 210$

```python
from itertools import permutations

num = '4445566'

a = set(permutations(num))
len(a)
```

```python
sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2))
```

### Пример 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

a). Рассмотрим событие $A$ – получение слова «ТОР»

$P(A) = \frac{1}{A_{6}^5} = \frac{1}{6 \cdot 5 \cdot 4} = \frac{1}{120}$

```python
from itertools import permutations

num = 'ТЕОРИЯ'

a = set(permutations(num, 3))
round(1 / len(a), 4)
```

```python
round(1 / sc.perm(6, 3), 4)
```

б). Рассмотрим событие $B$ – получение слова «ТЕОРИЯ».

$P(B) = \frac{1}{P_6} = \frac{1}{6!} = \frac{1}{720}$

```python
from itertools import permutations

num = 'ТЕОРИЯ'

a = list(permutations(num))
round(1 / len(a), 4)
```

```python
round(1 / sc.factorial(6), 4)
```

### Пример 1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

Рассмотрим событие $B$ – получение слова «АНАНАС».

$P(B) = \frac{P_3 \cdot P_2}{P_6} = \frac{3!2!}{6!} = \frac{1}{60}$

```python
round(sc.factorial(3) * sc.factorial(2) / sc.factorial(6), 4)
```

### Пример 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

a). Пусть событие А - никакие два лица одного пола не сядут рядом.

$P(A) = \frac{2(5!)^2}{10!} = \frac{2 \cdot 120^2}{3628800} = 0,00794$

```python
round(2 * (sc.factorial(5))**2 / sc.factorial(10), 5)
```

б). Пусть событие В - супруги, образующие пять супружеских пар, сядут рядом.

$P(B) = \frac{5!2^5}{10!} = \frac{120 \cdot 32}{3628800} = 0,00106$

```python
round(sc.factorial(5) * 2**5 / sc.factorial(10), 5)
```

### Пример 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

a). Пусть событие А - пассажиры попали в два купе.

$P(A) = \frac{C_9^2 \cdot C_8^7}{C_{36}^7} = \frac{36 \cdot 8}{8347680} = 0,0000345$

```python
round(sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7), 7)
```

б). Пусть событие В - пассажиры попали в семь купе.

$P(B) = \frac{C_9^7 \cdot 4^7}{C_{36}^7} = \frac{36 \cdot 16384}{8347680} = 0,07066$

```python
round(sc.comb(9, 7) * 4**7 / sc.comb(36, 7), 5)
```

в). Пусть событие D - пассажиры попали в три купе.

$P(D) = \frac{C_9^3(6 C_4^4 C_4^2 C_4^1 + 3 C_4^3 C_4^3 C_4^1 + 3 C_4^3 C_4^2 C_4^2)}{C_{36}^7} = \frac{64512}{8347680} = 0,00773$

```python
round(sc.comb(9, 3) * (6*sc.comb(4, 4)*sc.comb(4, 2)*sc.comb(4, 1) + 3*sc.comb(4, 3)*sc.comb(4, 3)*sc.comb(4, 1) + 3*sc.comb(4, 3)*sc.comb(4, 2)*sc.comb(4, 2)) / sc.comb(36, 7), 5)
```
